```python
import pandas as pd
import numpy as np
```

```python
cols_needed = [
    'case_status', 'decision_date', 'employer_name', 'employer_state',
    'employer_num_employees', 'wage_offer_from_9089', 'wage_offer_unit_of_pay_9089',
    'pw_soc_title', 'pw_level_9089', 'job_info_education', 'job_info_work_state'
]

df = pd.read_csv("us_perm_visas.csv", usecols=cols_needed, low_memory=False)

df['decision_date'] = pd.to_datetime(df['decision_date'], errors='coerce')
df['decision_year'] = df['decision_date'].dt.year
df['wage_offer_from_9089'] = pd.to_numeric(df['wage_offer_from_9089'], errors='coerce')
df['employer_num_employees'] = pd.to_numeric(df['employer_num_employees'], errors='coerce')

df = df.dropna(subset=['case_status', 'decision_year'])
df['decision_year'] = df['decision_year'].astype(int)

print(f"Rows after cleaning: {len(df)}")
df.to_parquet("perm_visas_clean.parquet", index=False)
df.head()
```
